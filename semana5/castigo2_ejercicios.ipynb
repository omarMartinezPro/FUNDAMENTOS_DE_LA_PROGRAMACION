```python
dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(km):
    total = 0
    dias_rodados = 0
    suma_rodados = 0
    mayor = 0
    indice_mayor = 0

    for i in range(len(km)):
        total += km[i]
        if km[i] > 0:
            dias_rodados += 1
            suma_rodados += km[i]
        if km[i] > mayor:
            mayor = km[i]
            indice_mayor = i

    promedio = round(suma_rodados / dias_rodados, 1)
    return total, dias_rodados, promedio, indice_mayor

total, dias_rodados, promedio, indice = resumen(km)
print("Total:", total, "km")
print("Días con rodada:", dias_rodados)
print("Promedio por día rodado:", promedio, "km")
print("Día más largo:", dias[indice], "("+str(km[indice])+" km)")
```

```python
abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

def expandir(mensaje):
    palabras = mensaje.lower().split()
    nuevas = []
    for palabra in palabras:
        nuevas.append(abrev.get(palabra, palabra))
    return " ".join(nuevas)

print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
```

```python
def rendimiento():
    while True:
        try:
            km = float(input("Kilómetros recorridos: "))
            litros = float(input("Litros cargados: "))
            resultado = km / litros
            return round(resultado, 1)
        except ValueError:
            print("Escribe solo números")
        except ZeroDivisionError:
            print("Los litros no pueden ser 0")

rend = rendimiento()
print("Rendimiento:", rend, "km/l")
if rend >= 20:
    print("Excelente")
elif rend >= 15:
    print("Normal")
else:
    print("Revisa tu moto")
```

```python
sala = [["X", "L", "L"], ["L", "X", "L"], ["X", "X", "L"]]

def reservar(sala):
    while True:
        try:
            fila = int(input("Fila: "))
            columna = int(input("Columna: "))
            if sala[fila][columna] == "X":
                print("Ocupado, elige otro")
            else:
                sala[fila][columna] = "X"
                return True
        except ValueError:
            print("Escribe números")
        except IndexError:
            print("Ese asiento no existe")

reservar(sala)
print("¡Reservado!")
for fila in sala:
    for asiento in fila:
        print(asiento, end=" ")
    print()
```

```python
def evaluar(pw):
    fallas = []
    tiene_numero = False
    tiene_mayuscula = False
    tiene_espacio = False

    if len(pw) < 8:
        fallas.append("mínimo 8 caracteres")
    for caracter in pw:
        if caracter.isdigit():
            tiene_numero = True
        if caracter.isupper():
            tiene_mayuscula = True
        if caracter == " ":
            tiene_espacio = True
    if not tiene_numero:
        fallas.append("falta un número")
    if not tiene_mayuscula:
        fallas.append("falta una mayúscula")
    if tiene_espacio:
        fallas.append("sin espacios")
    return fallas

for pw in ["metal", "Harley2008", "road glide 08", "KLR650moto"]:
    fallas = evaluar(pw)
    if len(fallas) == 0:
        print(pw+": segura")
    else:
        print(pw+":", fallas)
```

```python
alumnos = ["Bruce Dickinson", "Brian Dickinson", "Ana Lopez", "Andrea Lopez", "Angel Lopez"]

usados = []
def usuario(nombre):
    partes = nombre.lower().split()
    base = partes[0][0] + partes[1]
    candidato = base
    numero = 2
    while candidato in usados:
        candidato = base + str(numero)
        numero += 1
    usados.append(candidato)
    return candidato

for nombre in alumnos:
    print(nombre, "→", usuario(nombre))
```

```python
prohibidas = ["tonto", "menso", "baboso"]

def censurar(frase):
    palabras = frase.split()
    censuradas = 0
    for i in range(len(palabras)):
        if palabras[i].lower() in prohibidas:
            palabras[i] = "*" * len(palabras[i])
            censuradas += 1
    return " ".join(palabras), censuradas

frases = ["no seas menso y pásame el código", "Tonto el que no haga commit", "todo bien profe"]
for frase in frases:
    resultado, cantidad = censurar(frase)
    print(resultado, " ("+str(cantidad)+" censuradas)")
```

```python
autos = [("ABC123", 9, 11), ("XYZ789", 8, 15), ("MOT650", 10, 11), ("QRO001", 13, 13)]

def cobro(entrada, salida):
    horas = salida - entrada
    if horas <= 1:
        return 20
    else:
        return 20 + (horas - 1) * 15

cobros = {}
for placa, entrada, salida in autos:
    importe = cobro(entrada, salida)
    horas = salida - entrada
    cobros[placa] = importe
    print(placa+":", horas, "h → $"+str(importe))

total = 0
for importe in cobros.values():
    total += importe
print("Total del día: $"+str(total))
```

```python
partidos = (("Pumas", "Tigres", 2, 1), ("América", "Pumas", 0, 0), ("Tigres", "América", 3, 2),
            ("Pumas", "América", 1, 2), ("Tigres", "Pumas", 1, 1))

puntos = {}
for local, visitante, goles_local, goles_visitante in partidos:
    if local not in puntos:
        puntos[local] = 0
    if visitante not in puntos:
        puntos[visitante] = 0
    if goles_local > goles_visitante:
        puntos[local] += 3
    elif goles_local < goles_visitante:
        puntos[visitante] += 3
    else:
        puntos[local] += 1
        puntos[visitante] += 1

campeon = ""
mayor = -1
for equipo, puntaje in puntos.items():
    if puntaje > mayor:
        mayor = puntaje
        campeon = equipo
print(puntos)
print("Campeón:", campeon, "con", mayor, "puntos")
```

```python
nombres = ("Ana", "Luis", "Mía", "Beto")
cal = [[85, 90, 78, 92], [60, 75, 70, 68], [95, 98, 100, 97], [70, 65, 80, 75]]

def promedio(lista):
    suma = 0
    for numero in lista:
        suma += numero
    return suma / len(lista)

promedios_alumnos = []
for i in range(len(cal)):
    prom = promedio(cal[i])
    promedios_alumnos.append(prom)
    print(nombres[i]+":", prom)

for columna in range(len(cal[0])):
    parcial = []
    for fila in cal:
        parcial.append(fila[columna])
    print("Parcial", columna + 1, ":", promedio(parcial))

mejor_indice = 0
for i in range(len(promedios_alumnos)):
    if promedios_alumnos[i] > promedios_alumnos[mejor_indice]:
        mejor_indice = i
print("Mejor alumno:", nombres[mejor_indice])
```

```python
playlist = ["Master of Puppets - Metallica - 8:35", "Holy Wars - Megadeth - 6:32",
            "Fade to Black - Metallica - 6:57", "The Trooper - Iron Maiden - 4:10",
            "Hangar 18 - Megadeth - 5:14", "Aces High - Iron Maiden - 4:31"]

def a_segundos(tiempo):
    minutos, segundos = tiempo.split(":")
    return int(minutos) * 60 + int(segundos)

def formato(seg):
    minutos = seg // 60
    segundos = seg % 60
    if segundos < 10:
        return f"{minutos}:0{segundos}"
    else:
        return f"{minutos}:{segundos}"

artistas = {}
for linea in playlist:
    partes = linea.split(" - ")
    cancion, artista, tiempo = partes
    segundos = a_segundos(tiempo)
    if artista not in artistas:
        artistas[artista] = 0
    artistas[artista] += segundos

total = 0
artista_mayor = ""
tiempo_mayor = -1
for artista, segundos in artistas.items():
    print(artista+":", formato(segundos))
    total += segundos
    if segundos > tiempo_mayor:
        tiempo_mayor = segundos
        artista_mayor = artista
print("Más tiempo:", artista_mayor)
print("Total:", formato(total))
```
