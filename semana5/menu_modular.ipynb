```python
# MENÚ MODULAR: TUPLAS, DICCIONARIOS, EXCEPCIONES Y STRINGS
# Alumno: Omar Asael Martinez Chaparro.
# Matrícula: Al07283164
# Descripcion: Mi codigo realizado durante esta semana consiste en un menú modular que permite practicar tuplas, diccionarios, excepciones y strings. Cada opción llama a una función diferente y el menú continúa hasta que se selecciona la opción 5 para salir.


# PARTE A

def sumar_tupla(tupla):
    suma = 0

    for numero in tupla:
        suma = suma + numero

    return suma


def seccion_tuplas():
    numeros = (5, 10, 15, 20, 25)

    print("\nTupla original:", numeros)
    print("Tercer elemento:", numeros[2])

    numero1 = int(input("Ingresa el primer número: "))
    numero2 = int(input("Ingresa el segundo número: "))

    nueva_tupla = numeros + (numero1, numero2)

    print("Nueva tupla:", nueva_tupla)

    lista_numeros = list(nueva_tupla)
    lista_numeros.sort()

    print("Lista ordenada:", lista_numeros)

    resultado = sumar_tupla(nueva_tupla)

    print("Suma de la nueva tupla:", resultado)

# PARTE B

def buscar_telefono(contactos, nombre):
    if nombre in contactos:
        return contactos[nombre]
    else:
        return None


def seccion_diccionarios():
    contactos = {
        "Omar": "4481130317",
        "Adan": "4442345678",
        "Gustavo": "4443456789"
    }

    print("\nContactos registrados:")

    nombre_nuevo = input("Ingresa el nombre del nuevo contacto: ")
    telefono_nuevo = input("Ingresa su teléfono: ")

    contactos[nombre_nuevo] = telefono_nuevo

    print("\nLista de nombres:")

    for nombre in contactos:
        print(nombre)

    nombre_buscar = input("\n¿Qué contacto deseas buscar?: ")

    telefono = buscar_telefono(contactos, nombre_buscar)

    if telefono != None:
        print("Teléfono de", nombre_buscar, ":", telefono)
    else:
        print(nombre_buscar, "no está en tus contactos")


# PARTE C

def seccion_excepciones():
    print("\nIngresa dos números enteros")

    try:
        numero1 = int(input("Primer número: "))
        numero2 = int(input("Segundo número: "))

        suma = numero1 + numero2

        print("Suma:", suma)

        division = numero1 / numero2

        print("División:", division)

    except ValueError:
        print("Error: debes ingresar números enteros, no letras ni valores vacíos.")

    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero.")


# PARTE D

def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)


def seccion_strings():
    mensaje = "La programación es divertida y muy interesante"

    print("\nMensaje original:", mensaje)

    print("Longitud del mensaje:", len(mensaje))

    print("Mensaje en mayúsculas:", mensaje.upper())

    mensaje_nuevo = mensaje.replace("divertida", "importante")

    print("Mensaje reemplazado:", mensaje_nuevo)

    cantidad = contar_palabras(mensaje)

    print("Número de palabras:", cantidad)


# PARTE E

def menu():
    while True:
        print("\n========== MENU PRINCIPAL ==========")
        print("1. Tuplas")
        print("2. Diccionarios")
        print("3. Excepciones")
        print("4. Strings")
        print("5. Salir")

        opcion = input("Elige una opción: ")

        if opcion == "1":
            seccion_tuplas()

        elif opcion == "2":
            seccion_diccionarios()

        elif opcion == "3":
            seccion_excepciones()

        elif opcion == "4":
            seccion_strings()

        elif opcion == "5":
            print("Programa terminado. ¡Hasta luego!")
            break

        else:
            print("Opción inválida")


menu()






```
