```python
#TABLA DE PITÁGORAS EN MATRIZ
# Alumno: Omar Asael Martinez Chaparro.
# Matrícula: Al07283164
# Descripcion: Mi programa desarrollado en esta semana4 es una tabla de Pitágoras de 10 × 10 utilizando una matriz y ciclos for. Permite mostrar la tabla de manera ordenada y consultar multiplicaciones mediante una función que obtiene los resultados directamente de la matriz, sin realizar nuevamente la multiplicación.
# Construcción de la tabla de Pitágoras
tabla = []

for renglon in range(1, 11):
    fila = []
    for columna in range(1, 11):
        fila.append(renglon * columna)
    tabla.append(fila)

print("Tabla de Pitágoras de 10 x 10")


def imprimir_tabla(tabla):
    for fila in tabla:
        for numero in fila:
            print(numero, end="\t")
        print()


# Consultar un producto

def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]


# Mostrar la tabla
imprimir_tabla(tabla)

# Consultar productos continuamente

while True:
    # Entrada para el primer factor
    entrada_renglon = input("Ingresa el primer factor (1 al 10) o escribe fin: ").strip()

    if entrada_renglon.lower() == "fin":
        print("Programa terminado")
        break

    try:
        renglon = int(entrada_renglon)
        if not (1 <= renglon <= 10):
            print("Aviso: el número debe estar entre 1 y 10.")
            continue
    except ValueError:
        print("Aviso: Debes ingresar un número entero válido o escribir 'fin'.")
        continue

    # Entrada para el segundo factor
    while True:
        entrada_columna = input("Ingresa el segundo factor (1 al 10) o escribe fin: ").strip()
        if entrada_columna.lower() == "fin":
            print("Programa terminado")
            break
        try:
            columna = int(entrada_columna)
            if 1 <= columna <= 10:
                break
            else:
                print("Aviso: el número debe estar entre 1 y 10.")
        except ValueError:
            print("Aviso: Debes ingresar un número entero válido.")

    if entrada_columna.lower() == "fin":
        break

    resultado = consultar_producto(tabla, renglon, columna)
    print(renglon, "x", columna, "=", resultado)
```
