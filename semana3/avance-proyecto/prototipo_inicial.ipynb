```python
# Alumno: Omar Asael Martinez Chaparro.
# Matrícula: Al07283164
# ==========================================
# RESTAURANTE MAR Y LUZ
# ==========================================

# 1. PRECIOS DE LOS PLATILLOS
PRECIO1 = 120.0  # Carne asada
PRECIO2 = 120.0  # Pechuga empanizada
PRECIO3 = 110.0  # Pancita con patita de res
PRECIO4 = 130.0  # Menudo de borrego
PRECIO5 = 150.0  # Caldo de guajolote
PRECIO6 = 150.0  # Mole de guajolote
PRECIO7 = 110.0  # Chiles rellenos
PRECIO8 = 110.0  # Carne de puerco

# Listas de platillos y precios
nombres_menu = [
    "Carne asada",
    "Pechuga empanizada",
    "Pancita con patita de res",
    "Menudo de borrego",
    "Caldo de guajolote",
    "Mole de guajolote",
    "Chiles rellenos",
    "Carne de puerco"
]

precios_menu = [
    PRECIO1, PRECIO2, PRECIO3, PRECIO4,
    PRECIO5, PRECIO6, PRECIO7, PRECIO8
]


# 2. PRESENTACION Y SELECCION DE MESA
print("===============================")
print("     RESTAURANTE MAR Y LUZ")
print("===============================")
print("Mesas disponibles: 1 al 8")

while True:
    try:
        mesa = int(input("Ingrese el numero de mesa: "))

        if 1 <= mesa <= 8:
            break
        else:
            print("Mesa no valida. Seleccione una mesa del 1 al 8.")

    except ValueError:
        print("Ingrese un numero entero valido.")

# 3. VARIABLES PARA GUARDAR LOS PEDIDOS
nombres = []
cantidades = []
precios = []
importes = []

pedido_activo = True

# 4. MENU PRINCIPAL
while pedido_activo:

    print("\nMENU PRINCIPAL")
    print("1. Agregar platillo")
    print("2. Consultar pedido")
    print("3. Finalizar cuenta")

    try:
        opcion = int(input("Seleccione una opcion: "))
    except ValueError:
        print("Opcion no valida. Ingrese un numero del 1 al 3.")
        continue

    # OPCION 1: AGREGAR PLATILLO
    if opcion == 1:

        if len(nombres) < 8:

            print("\nMENU DE PLATILLOS")

            for i in range(8):
                print(i + 1, ".", nombres_menu[i],
                      "- $", format(precios_menu[i], ".2f"))

            try:
                platillo = int(input("Seleccione el platillo (1-8): "))
            except ValueError:
                print("Platillo no valido.")
                continue

            if 1 <= platillo <= 8:

                while True:
                    try:
                        cantidad = int(input("Ingrese la cantidad: "))

                        if cantidad > 0:
                            break
                        else:
                            print("La cantidad debe ser mayor que cero.")

                    except ValueError:
                        print("Ingrese una cantidad entera valida.")

                nombre = nombres_menu[platillo - 1]
                precio = precios_menu[platillo - 1]
                importe = precio * cantidad

                nombres.append(nombre)
                cantidades.append(cantidad)
                precios.append(precio)
                importes.append(importe)

                print("\nPlatillo registrado correctamente.")
                print("Importe del platillo: $", format(importe, ".2f"))
                print("Una vez enviado a cocina, no se puede modificar.")

            else:
                print("Seleccione un platillo del 1 al 8.")

        else:
            print("Se alcanzo el limite de ocho registros.")
            print("Consulte el pedido o finalice la cuenta.")

    # OPCION 2: CONSULTAR PEDIDO
    elif opcion == 2:

        print("\n===============================")
        print("       PEDIDO DE LA MESA", mesa)
        print("===============================")

        if len(nombres) == 0:
            print("No hay platillos registrados.")

        else:
            total = 0.0

            for i in range(len(nombres)):
                print(
                    "Registro", i + 1,
                    "|", nombres[i],
                    "| Cantidad:", cantidades[i],
                    "| Precio: $", format(precios[i], ".2f"),
                    "| Importe: $", format(importes[i], ".2f")
                )

                total = total + importes[i]

            print("--------------------------------")
            print("TOTAL ACTUAL: $", format(total, ".2f"))

    # OPCION 3: FINALIZAR CUENTA
    elif opcion == 3:

        if len(nombres) == 0:
            print("No hay platillos registrados.")
            print("Agregue al menos un platillo antes de finalizar.")

        else:
            pedido_activo = False
            print("Finalizando cuenta...")

    else:
        print("Opcion no valida. Seleccione del 1 al 3.")

# 5. CUENTA FINAL
total = 0.0

print("\n===============================")
print("          CUENTA FINAL")
print("===============================")
print("Mesa:", mesa)

for i in range(len(nombres)):
    print(
        "Registro", i + 1,
        "|", nombres[i],
        "| Cantidad:", cantidades[i],
        "| Precio: $", format(precios[i], ".2f"),
        "| Importe: $", format(importes[i], ".2f")
    )

    total = total + importes[i]

print("--------------------------------")
print("TOTAL DE LA MESA: $", format(total, ".2f"))

# 6. FORMA DE PAGO
print("\n¿COMO DESEA PAGAR?")
print("1. Pago junto")
print("2. Pago separado por platillo")

while True:
    try:
        forma_pago = int(input("Seleccione 1 o 2: "))

        if forma_pago == 1 or forma_pago == 2:
            break
        else:
            print("Seleccione 1 para pago junto o 2 para separado.")

    except ValueError:
        print("Ingrese un numero entero valido.")

# 7. PAGO JUNTO
if forma_pago == 1:

    print("\n===============================")
    print("             PAGO JUNTO")
    print("===============================")
    print("Una sola cuenta para toda la mesa.")
    print("TOTAL A PAGAR: $", format(total, ".2f"))

# 8. PAGO SEPARADO POR PLATILLO
else:

    print("\n===============================")
    print("      PAGO SEPARADO")
    print("===============================")
    print("Cada persona pagara los platillos que consumio.")

    print("\nREGISTROS DISPONIBLES")

    for i in range(len(nombres)):
        print(
            "Registro", i + 1,
            "|", nombres[i],
            "| Cantidad:", cantidades[i],
            "| Importe: $", format(importes[i], ".2f")
        )

    # Solicitar el numero de personas
    while True:
        try:
            personas = int(input("\n¿Cuantas personas pagaran por separado? "))

            if personas > 0 and personas <= len(nombres):
                break
            else:
                print(
                    "El numero de personas debe ser mayor que cero "
                    "y no superar el numero de registros."
                )

        except ValueError:
            print("Ingrese un numero entero valido.")

    # Repetir la asignacion si quedan registros sin asignar
    while True:

        registros_asignados = []
        pagos = []

        # Solicitar los registros de cada persona
        for persona in range(1, personas + 1):

            while True:
                entrada = input(
                    "\nPersona " + str(persona)
                    + ", Ingrese el numero de pedido que consumiste separado con comas: "
                )

                try:
                    registros = []

                    for dato in entrada.split(","):
                        registros.append(int(dato.strip()))

                    # Verificar que se ingresen registros
                    if len(registros) == 0:
                        print("Debe ingresar al menos un registro.")
                        continue

                    # Evitar repetir registros en la misma persona
                    if len(set(registros)) != len(registros):
                        print("No repita el mismo registro.")
                        continue

                    # Verificar que los registros existan
                    validos = True

                    for numero in registros:
                        if numero < 1 or numero > len(nombres):
                            validos = False

                    if not validos:
                        print("Uno de los registros no existe.")
                        continue

                    # Evitar cobrar un registro a dos personas
                    repetido = False

                    for numero in registros:
                        if numero in registros_asignados:
                            repetido = True

                    if repetido:
                        print("Un registro ya fue asignado a otra persona.")
                        continue

                    break

                except ValueError:
                    print("Ingrese el numero de pedido que consumiste separado con comas. Ejemplo: 1,3")

            # Calcular el pago de esta persona
            subtotal = 0.0

            for numero in registros:
                subtotal = subtotal + importes[numero - 1]

            registros_asignados.extend(registros)
            pagos.append(subtotal)

        # Comprobar que todos los registros esten asignados
        if len(registros_asignados) == len(nombres):
            break
        else:
            print("\nFaltan registros por asignar.")
            print("Se solicitara nuevamente la distribución.")
            print("Cada platillo debe asignarse una sola vez.")

    # Mostrar los pagos individuales
    print("\n===============================")
    print("       RESUMEN DE PAGOS")
    print("===============================")

    total_pagos = 0.0

    for persona in range(personas):
        print(
            "Persona", persona + 1,
            "debe pagar: $", format(pagos[persona], ".2f")
        )

        total_pagos = total_pagos + pagos[persona]

    print("--------------------------------")
    print("Total de la mesa: $", format(total, ".2f"))
    print("Suma de pagos individuales: $", format(total_pagos, ".2f"))

    if round(total_pagos, 2) == round(total, 2):
        print("Comprobacion correcta: los pagos coinciden con la cuenta.")
    else:
        print("Error: los pagos no coinciden con el total.")


print("\nGracias por visitar el Restaurante Mar y Luz.")
print("¡Vuelva pronto!")
```
