```python
# TAQUILLA DEL MUSEO.
# Alumno: Omar Asael Martinez Chaparro.
# Matrícula: Al07283164
# Descripción: Este programa automatiza el cobro de boletos en la taquilla de un
# museo según la edad y categoría del visitante, aplicando descuentos
# automáticos o por menú, y generando un corte de caja al finalizar.

# Constantes
PRECIO_NIÑO = 30.00
PRECIO_ADULTO = 45.00
DESCUENTO_ESTUDIANTE = 0.10
DESCUENTO_PROFESOR = 0.10
DESCUENTO_ADULTO_MAYOR = 0.12

# Contadores y acumuladores
visitantes = 0
gratis = 0
estudiantes = 0
profesores = 0
adultos_mayores = 0

total_descuentos = 0.00
total_pagar = 0.00

print("TAQUILLA DEL MUSEO")

while True:

    entrada = input("Edad del visitante o escribe fin para cerrar: ")

    if entrada == "fin":
        break

    # Validar que la entrada sea un número entero
    try:
        edad = int(entrada)
    except ValueError:
        print("Entrada no válida. Por favor, ingresa un número entero o 'fin'.")
        continue

    # Validar edad
    if edad < 0 or edad > 110:
        print("Edad no valida. No se cobra.")
        continue

    # Calcular precio y descuento
    descuento = 0.00

    if edad < 3:
        precio = 0.00
        tipo = "Gratis"
        gratis = gratis + 1

    elif edad < 18:
        precio = PRECIO_NIÑO
        tipo = "Menor de edad"

    elif edad >= 60:
        precio = PRECIO_ADULTO
        tipo = "Adulto mayor"
        descuento = precio * DESCUENTO_ADULTO_MAYOR
        adultos_mayores = adultos_mayores + 1

    else:
        precio = PRECIO_ADULTO

        print("1) Estudiante")
        print("2) Profesor")
        print("3) Ninguno")

        opcion = input("Selecciona el tipo de visitante: ")

        if opcion == "1":
            tipo = "Estudiante"
            descuento = precio * DESCUENTO_ESTUDIANTE
            estudiantes = estudiantes + 1

        elif opcion == "2":
            tipo = "Profesor"
            descuento = precio * DESCUENTO_PROFESOR
            profesores = profesores + 1

        else:
            tipo = "Ninguno"

    # Calcular el pago
    pago = precio - descuento

    # Actualizar contadores y acumuladores
    visitantes = visitantes + 1
    total_descuentos = total_descuentos + descuento
    total_pagar = total_pagar + pago

    # Mostrar el boleto
    print("Visitante:", visitantes,
          "| Edad:", edad,
          "| Tipo:", tipo,
          "| Base: $", format(precio, ".2f"),
          "| Descuento: $", format(descuento, ".2f"),
          "| Paga: $", format(pago, ".2f"))

# Corte de caja
print("\n=== CORTE DE CAJA ===")
print("Visitantes atendidos:", visitantes)
print("Boletos gratis:", gratis)
print("Boletos de estudiante:", estudiantes)
print("Boletos de profesor:", profesores)
print("Boletos de adulto mayor:", adultos_mayores)
print("Total en descuentos: $", format(total_descuentos, ".2f"))
print("TOTAL RECAUDADO: $", format(total_pagar, ".2f"))
```
